##**Introdução**
A análise de dados geográficos tem se tornado cada vez mais importante para compreender a distribuição de serviços públicos e auxiliar na tomada de decisões. Na área da saúde, a visualização espacial das Unidades Básicas de Saúde (UBS) permite identificar padrões de concentração e cobertura em diferentes regiões do Brasil.

Neste notebook, foram utilizadas as bibliotecas Pandas e GeoPandas para realizar o tratamento, análise e visualização dos dados disponibilizados em formato CSV. Foram realizadas análises exploratórias, contagem de unidades por estado e representação geográfica das UBS utilizando coordenadas de latitude e longitude.

O objetivo principal deste trabalho é explorar os dados de maneira visual e estatística, demonstrando a aplicação de ferramentas de análise de dados em Python no contexto da saúde pública.

# **IMPORTS E TRATAMENTO DE DADOS**

```python
#Imports necessarios
!pip install geopandas geobr
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import geobr as geobr
```

```python
#Criando dataframe com os dados do csv
df = pd.read_csv('/content/Unidades_Basicas_Saude-UBS.csv', sep=';')
```

```python
df.head()

df.info()

df.describe()
```

```python
#O csv veio com , aqui troco para ponto

df['LONGITUDE'] = (
    df['LONGITUDE']
    .astype(str)
    .str.replace(',', '.')
    .astype(float)
)
```

```python
#Criando um GeoDataFrame para fazermos uso do GeoPandas
gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(
        df['LONGITUDE'],
        df['LATITUDE']
    )
)
```

# **GRÁFICO DE UBS POR UF**

```python
#criando um gráfico mostrando a quantidade de UBS por UF
#Lembrando que no csv UF veio como uma id
ubs_por_uf = df['UF'].value_counts()

ubs_por_uf.plot(kind='bar', figsize=(12,6))

plt.title('Quantidade de UBS por UF')
plt.show()
```

# **MAPA DE DISTRIBUIÇÃO DE UBS**

```python
#Plotando um mapa com a distribuição de UBS no Brasil
gdf.plot(figsize=(10,10), markersize=1)

plt.title('Distribuição das UBS no Brasil')
plt.show()
```

# **MAPA DE CONCENTRAÇÃO DE UBS POR UF**

```python
#Contando UBS por estado

ubs_por_uf = (
    df['UF']
    .value_counts()
    .reset_index()
)

ubs_por_uf.columns = ['abbrev_state', 'quantidade']

# Garantir que a coluna seja string
ubs_por_uf['abbrev_state'] = (
    ubs_por_uf['abbrev_state']
    .astype(str)
)
```

```python
#Para conseguir fazer,crio um mapping das ufs, corrigindo o problema lá de cima sobre uf ser codigo
uf_mapping = {
    11: 'RO', 12: 'AC', 13: 'AM', 14: 'RR', 15: 'PA', 16: 'AP', 17: 'TO', # Norte
    21: 'MA', 22: 'PI', 23: 'CE', 24: 'RN', 25: 'PB', 26: 'PE', 27: 'AL', 28: 'SE', 29: 'BA', # Nordeste
    31: 'MG', 32: 'ES', 33: 'RJ', 35: 'SP', # Sudeste
    41: 'PR', 42: 'SC', 43: 'RS', # Sul
    50: 'MS', 51: 'MT', 52: 'GO', 53: 'DF'  # Centro-Oeste
}


display(ubs_por_uf.head())
```

```python
#Baixando mapa do
brasil = geobr.read_state()
```

```python

# Garantir no mapa também
brasil['abbrev_state'] = (
    brasil['abbrev_state']
    .astype(str)
)
#Juntando mapa com os dados que temos
mapa = brasil.merge(
    ubs_por_uf,
    on='abbrev_state'
)
mapa = brasil.merge(
    ubs_por_uf,
    on='abbrev_state'
)
```

```python
fig, ax = plt.subplots(figsize=(15,10))

mapa.plot(
    column='quantidade',
    cmap='OrRd',
    linewidth=0.8,
    edgecolor='black',
    legend=True,
    ax=ax
)

ax.set_aspect('auto')

plt.title('Quantidade de UBS por Estado')

plt.axis('off')

plt.show()
```
